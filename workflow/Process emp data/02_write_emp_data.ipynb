```python

dbutils.widgets.text("dept", "","Departmet Name")
```

```python
dept = dbutils.widgets.get("dept")
```

```python
print(dept)
```

```python
df=spark.read.format("csv").option("header", True).load("/databricks-datasets/retail-org/company_employees/company_employees.csv")
```

```python
display(df)
```

```python
df_filtered=df.where(f"upper(department)=upper('{dept}') and active_record='1'")  
display(df_filtered )
```

```python
count =df_filtered.count()
if count  > 0:
    df_filtered.write.mode("overwrite").saveAsTable(f"bronze.dept_{dept}")
    print("data written for {dept}")
    display(spark.table(f"bronze.dept_{dept}"))
else:
    print("no data for {dept}")
```

```python
dbutils.notebook.exit(count)
```
