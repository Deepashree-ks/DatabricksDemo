```python
input_date=dbutils.widgets.get("input_date")
print(input_date)
```

```python
spark.conf.set("spark.sql.legacy.timeParserPolicy","LEGACY")

input_day1=spark.sql(f"""
                    select date_format(to_timestamp('{input_date}','yyyy-MM-dd'),'E')
                    """).collect()[0][0]
print(input_day1)
```

```python
dbutils.jobs.taskValues.set("input_day", input_day1)
```
